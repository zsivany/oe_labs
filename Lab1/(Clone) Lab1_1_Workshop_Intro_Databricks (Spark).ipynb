##Dataframe operation types
#

The dataframe processing (key concept in Spark) is the `lazy evaluation`. What does it mean?
The operations have two main groups: 

 - `Action`:
   * An action provide the ability to return previews or specify physical execution plans for how logic will map to data. 
   * In other words: Actions are commands that are computed by Spark right at the time of their execution. 
   * They consist of running all of the previous transformations in order to get back an actual result. 
   * An action is composed of one or more jobs which consists of tasks that will be executed by the workers in parallel where possible
 - `Transformation`:
   * They eventually return another DataFrame.
   * They are immutable - that is each instance of a DataFrame cannot be altered once it's instantiated.
   * Are classified as either a Wide or Narrow operation
      
 - A `wide` transformation requires sharing data across workers (e.g.: join, groupby, dropDuplicate) -> need to shuffle
 - A `narrow` transformation can be applied per partition/worker with no need to share or shuffle data to other workers (e.g.: select, where, union, etc.)

### Key Considerations When Choosing
* Simplicity vs. Control: Do you need a hands-off, fully managed experience or detailed control over your compute environment? 
* Cost Management: Are you running short, frequent jobs (where Serverless might be cheaper) or long, stable jobs (where Classic might be better)? 
* Workload Type: Serverless is great for simplicity and speed, while Classic is better for workloads that need extensive customization. 
* Technical Expertise: Serverless requires less infrastructure management expertise, whereas Classic benefits from users who want to manage their own resources.

##Magic Commands
* Magic Commands are specific to the Databricks notebooks
* They are very similar to Magic Commands found in comparable notebook products (language agnostic capability)
* These are built-in commands that do not apply to the notebook's default language
* A single percent (%) symbol at the start of a cell identifies a Magic Commands
* E.g.: `scala`, `r`, `python`, `sql`, `sh`, `run`, `md`, etc..
