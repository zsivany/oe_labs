```sql
SELECT
pipeline_event_id,
expectation.name, 
expectation.dataset, 
expectation.passed_records, 
expectation.failed_records,
( to_number(expectation.failed_records, '999999')/to_number(expectation.passed_records, '999999')*100) as failed_ratio,

*
FROM system.lakeflow_pipeline_events_preview.pipeline_events
LATERAL VIEW EXPLODE(
    FROM_JSON(
        details:flow_progress.data_quality.expectations::string,
        'ARRAY<STRUCT<
            name: STRING,
            dataset: STRING,
            passed_records: STRING,
            failed_records: STRING
        >>'
    )
) AS expectation
where pipeline_id = 'ef3c46af-2ea8-4858-b49a-40e59e5a02be'
--and pipeline_event_id = '68458670-dec0-11f1-afc4-3a44a627a170'
and event_type = 'flow_progress'
limit 10
```
